How to find all the local maxima (or peaks) in a numeric series?

```python
ser = pd.Series([2, 10, 3, 4, 9, 10, 2, 7, 3])
```

```python
n = ser.shape[0]
```

```python
ans = []
for i in range(n-1):
  if i==0:
    continue
  if ser[i]>ser[i-1] and ser[i]>ser[i+1]:
    ans.append(i)
```

```python
ans
```

```python
ser[ans]
```
